# Data Type

Data Types:
1. Integer (whole numbers even negative or positive)
2. Float (decimal numbers)
3. (bonus: complex imaginary number (j))
4. Boolean (False, True)

 Sequence Data Type (string, list, tuple)
5. String (array byte unicode) put string between one, two or three (apostrof) quotation mark '', "", """"""
6. List (ordered, changeable, allows duplicate members) put list between square brackets []
7. Tuple (ordered, unchangeable, allows duplicate members) put tuple between parentheses ()
8. Sets (unordered, unchangeable, no duplicate members) put set between curly brackets {}
9. Dictionary (unordered, changeable, no duplicate members) put dictionary between curly brackets {} with key and value pair

How to check data types: using type() method

### indexing 
indexing is for accessing individual characters in a string
index start with 0 

```python
a = "Hello, World!"
```

```python
print(a[1]) # indexing starts from 0, so a[1] will return the second character in the string, which is 'e'
print(a[7]) # a[7] will return the eighth character in the string, which is 'W'
```

```python
print(a[:5]) # slicing, a[:5] will return the first 5 characters in the string, which is 'Hello'
print(a[7:]) # a[7:] will return the characters from the eighth character
```

```python
print(a[-]) # negative indexing, a[-3] will return the third character from the end of the string, which is 'r'
```

```python
print(a[2:5]) # a[2:5] will return the characters from the third character to the fifth character, which is 'llo'   
```

```python
a*3 # string repetition, a*3 will return the string 'Hello, World!' repeated 3 times    
```

```python
a + a # string concatenation, a + a will return the string 'Hello, World!Hello, World!' which is the result of concatenating the string 'Hello, World!' with itself
```

```python
coffee[0] = 'luwak' # we can also change the value of an element in the list by using indexing, this will change the first element of the coffee list to 'luwak'
print(coffee) # this will print the updated coffee list, which is ['luwak',
```

```python
nested_list = ['luwak', ['nasi goreng', 'sate'], True] # this is a nested list, which contains a string 'luwak', another list ['nasi goreng', 'sate'], and a boolean value True
print(nested_list[1]) # this will print True, which is the third element of the nested list
print(nested_list[1][0]) # this will print 'nasi goreng', which is the first element of the second element of the nested list
```

### Sets
 sets is a collection of unique elements, it does not allow duplicate elements, it is unordered, it is mutable, we can add or remove elements from a set, we can use set() function to create a set

```python
points = {1, 2, 3, 4} # this is a set, which is an unordered collection of unique elements, this set has 4 elements, which are 1, 2, 3, and 4
```

```python
type(points)
```

```python
print(points) # this will print the set, which is {1, 2, 3, 4}
```

```python
points_log = {1, 2, 3, 4, 4, 4} # this set has duplicate elements, but the set will only keep unique elements, so the duplicate elements will be removed, this set will only have 4 unique elements, which are 1, 2, 3, and 4
print(points_log) # this will print the set, which is {1, 2, 3, 4}  
```

```python
alun_points_log = {1, 2, 3, 4, 5, 5, 5, 6, 7} # this set has duplicate elements, but the set will only keep unique elements, so the duplicate elements will be removed, this set will only have 5 unique elements, which are 1, 2, 3, 4, 5, 6, and 7
```

```python
print(points_log | alun_points_log) # this will print the union of the two sets, which is {1, 2, 3, 4, 5, 6, 7}
```

```python
print(points_log & alun_points_log) # this will print the intersection of the two sets, which is {1, 2, 3, 4}   
```

```python
print(points_log - alun_points_log) # this will print the difference of the two sets, which is an empty set because all elements in points_log are also in alun_points_log
```

```python
print(alun_points_log - points_log) # this will print the difference of the two sets, which is {5, 6, 7} because these elements are in alun_points_log but not in points_log    
```

### Dictionary
 dictionary is a collection of **key-value pairs**, where each key is unique and maps to a value, the keys in a dictionary must be immutable data types, such as strings, numbers, or tuples, while the values can be of any data type, including lists, sets, or even other dictionaries


```python
students_score = {'Alice': 85, 'Bob': 90, 'Charlie': 78} # this is a dictionary, which is an unordered collection of key-value pairs, this dictionary has 3 key-value pairs, which are 'Alice': 85, 'Bob': 90, and 'Charlie': 78
print(students_score) 
```

```python
my_data = {'name': 'Alice', 'age': 25, 'city': 'New York', 'favorite_color': ['blue', 'yellow']} # this is a dictionary, which is an unordered collection of key-value pairs, this dictionary has 4 key-value pairs, which are 'name': 'Alice', 'age': 25, 'city': 'New York', and 'favorite_color': ['blue', 'yellow']
print(my_data)
```

```python
type(my_data)
```

```python
my_data.values() # this will return the values of the dictionary, which is dict_values(['Alice', 25, 'New York', ['blue', 'yellow']])   
```

```python
my_data.keys() # this will return the keys of the dictionary, which is dict_keys(['name', 'age', 'city', 'favorite_color']) 
```

```python
my_data[0] # this will throw an error because dictionary is not ordered, we cannot access the elements of a dictionary by using indexing, we can only access the elements of a dictionary by using the keys, we cannot use indexing to access the values of a dictionary, we can only use the keys to access the values of a dictionary 
```

```python
my_data['age']
```

```python
my_data['age'] = '22' # we can change the value of a key in the dictionary by using indexing with the key, this will change the value of the 'age' key to '22'
print(my_data) # this will print the updated dictionary, which is {'name': 'Alice', 'age': '22', 'city': 'New York', 'favorite_color': ['blue', 'yellow']}
```

```python
my_data.update({'name': 'Alicia', 'favorite_color': ['red', 'green'], 'weight': 60}) # we can also change the value of a key in the dictionary by using the update() method, this will change the value of the 'name' key to 'Alicia' and the value of the 'favorite_color' key to ['red', 'green'] and also add a new key-value pair 'weight': 60 to the dictionary
print(my_data) # this will print the updated dictionary, which is {'name': 'Alicia', 'age': '22', 'city': 'New York', 'favorite_color': ['red', 'green'], 'weight': 60}
```
